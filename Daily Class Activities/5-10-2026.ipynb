```python
from sklearn.datasets import load_iris
from sklearn.svm import SVC
iris=load_iris()
x=iris.data[:,[0]]
y=iris.target
model=SVC()
model.fit(x,y)
print(model.predict(x[[50]]))
```

```python
from sklearn.datasets import load_iris
from sklearn.linear_model import SGDClassifier
iris=load_iris()
x=iris.data[:,[1]]
y=iris.target
model=SGDClassifier()
model.fit(x,y)
print(model.predict(x[[149]]))
```
